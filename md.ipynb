```python
import pandas as pd 
```

```python
from pathlib import Path
df = pd.read_csv(Path("data") / "spotify_data.csv")
```

```python
df
```

```python
df_sample = df.sample(n=min(5000, len(df)), random_state=42)
```

```python
feature_cols = ["danceability", "energy", "valence", "tempo", "acousticness", "liveness", "instrumentalness", "loudness", "speechiness"]
x = df_sample[feature_cols]
y= df_sample['track_genre']
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.ensemble import RandomForestClassifier
model = RandomForestClassifier(n_estimators=50, random_state=42)
model.fit(x_train, y_train)
```

```python
model.get_params()
```

```python
model.score(x_test, y_test)
```

```python
model.score(x_train, y_train)
```

```python
from sklearn.dummy import DummyClassifier
baseline = DummyClassifier(strategy="most_frequent")
baseline.fit(x_train, y_train)
```

```python
baseline.score(x_test, y_test)
```

```python
baseline.score(x_train, y_train)
```
